## Import Libraries

```python
import gymnasium as gym
import numpy as np
import matplotlib.pyplot as plt
```

## Task 1: Verify Installation

```python
print("Gymnasium Version:", gym.__version__)
```

## Task 2: Create and Initialize CartPole Environment

```python
# Create CartPole environment
env = gym.make("CartPole-v1")

# Reset environment
observation, info = env.reset()

# Display initial observation and environment info
print("Initial Observation:")
print(observation)

print("\nEnvironment Info:")
print(info)
```

## Task 3: Explore Observation and Action Spaces

```python
print("Observation Space:")
print(env.observation_space)

print("\nAction Space:")
print(env.action_space)

print("\nType of Observation Space:")
print(type(env.observation_space))

print("\nNumber of Possible Actions:")
print(env.action_space.n)
```

## Task 4: Execute a Random Agent

```python
# Reset environment
observation, info = env.reset()

done = False
step = 0
total_reward = 0

print("-" * 70)

while not done:

    # Select random action
    action = env.action_space.sample()

    # Execute action
    observation, reward, terminated, truncated, info = env.step(action)

    done = terminated or truncated

    step += 1
    total_reward += reward

    print(f"Step {step}")
    print(f"Action: {action}")
    print(f"Observation: {observation}")
    print(f"Reward: {reward}")
    print(f"Episode Finished: {done}")
    print("-" * 70)

print("Episode Completed!")
print("Total Steps:", step)
print("Cumulative Reward:", total_reward)

env.close()
```
